# Preprocesamiento y análisis de datos en Pandas

```python
import pandas as pd

# Distribución del gasto turístico y gasto medio diario realizado según continente de destino
gasto = pd.read_csv('https://www.ine.es/jaxiT3/files/t/csv_bdsc/58705.csv', sep=';')
gasto
```

```python
# Podemos ver en detalle los valores únicos en la columna Total
gasto.Total.unique()
```

```python
# Gasto total en todas las filas
print('Suma de totales:', gasto.Total.sum(), '\n')  # Sin ningún sentido
print('Total máximo:', gasto.Total.max(), '\n')
print('Total mínimo:', gasto.Total.min(), '\n')
gasto.Total.describe()
```

```python
# Problema 1: los totales son cadenas de texto con el punto (".") como separador de millares => eliminarlo
gasto['total_limpio'] = gasto.Total.apply(lambda x: x.replace('.', ''))
gasto
```

```python
# Problema 2: 'total_limpio' tiene algunos valores vacíos '' => eliminar esas filas
gasto[gasto['total_limpio'] == '']
# Hay varias filas con el total_limpio vacío
```

```python
gasto = gasto[gasto['total_limpio'] != '']
gasto
# Nos quedamos con las filas que tienen total_limpio no vacío
```

```python
# Problema 3: Algunas filas tienen valores decimales separados con coma
# (',') en 'total_limpio', que es algo muy español pero funciona mal 
# en Python => cambiar comas por puntos
gasto[gasto['total_limpio'].str.find(',') >= 0]
```

```python
gasto['total_limpio'] = gasto.total_limpio.apply(lambda x: x.replace(',', '.'))
gasto
```

```python
# Problema 4: 'total_limpio' es una columna de cadenas de texto, pasar a número
gasto['total_limpio'] = gasto['total_limpio'].apply(lambda x: float(x))
gasto
```

```python
# Gasto total en todas las filas
print('Suma de totales:', gasto.total_limpio.sum(), '\n')
print('Total máximo:', gasto.total_limpio.max(), '\n')
print('Total mínimo:', gasto.total_limpio.min(), '\n')
gasto.total_limpio.describe()
```
