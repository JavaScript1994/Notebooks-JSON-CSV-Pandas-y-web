# Web scraping

El **web scraping** es una técnica utilizada para extraer información de sitios web de manera automatizada. A través del web scraping, es posible obtener datos estructurados de páginas que normalmente están diseñadas para ser leídas por humanos. Esta técnica es muy útil para recolectar información cuando no existe una API oficial que proporcione los datos necesarios. El proceso de web scraping implica realizar solicitudes HTTP a una página web y luego analizar el contenido HTML para extraer los datos de interés. 

En Python, existen varias bibliotecas que facilitan el proceso de web scraping:

- **`requests`** (https://github.com/psf/requests): Permite realizar solicitudes HTTP de manera sencilla.
- **`BeautifulSoup`** (https://www.crummy.com/software/BeautifulSoup/bs4/doc/): Facilita el análisis y la extracción de datos de documentos HTML y XML.
- **`MechanicalSoup`** (https://github.com/MechanicalSoup/MechanicalSoup): biblioteca Python para automatizar algunas interacciones sencillas con páginas web.
- **`Selenium`** (https://selenium-python.readthedocs.io/) y **Playwright** (https://playwright.dev/): Permite interactuar de manera completa con páginas web dinámicas (que cargan contenido mediante JavaScript) simulando la interacción de un usuario en un navegador web.
- **`Scrapy`** (https://scrapy.org/): Un framework completo de scraping que permite extraer, procesar y almacenar datos de manera eficiente.

## Consideraciones Éticas y Legales

Es importante recordar que el web scraping debe hacerse de manera ética y legal. Antes de extraer datos de un sitio web, es recomendable revisar los términos de servicio del mismo y verificar si permite o prohíbe esta práctica. Además, muchas páginas web ofrecen APIs oficiales para acceder a sus datos de manera más controlada. Algunos aspectos importantes a considerar:

### 1. **Términos de Servicio (ToS)**
Muchos sitios web establecen en sus términos de servicio (ToS) si permiten o prohíben el web scraping. Los términos de servicio son acuerdos legales entre el usuario y el sitio web, y violarlos puede tener consecuencias legales.

### 2. **Ley de Protección de Datos**
En algunos casos, los datos extraídos mediante scraping pueden estar sujetos a leyes de protección de datos, como el **Reglamento General de Protección de Datos (RGPD)** en la Unión Europea. Si se recopilan datos personales, como nombres o direcciones de correo electrónico, los responsables del scraping deben asegurarse de cumplir con las leyes locales de protección de datos.

### 3. **Derechos de Autor y Propiedad Intelectual**
El contenido extraído mediante web scraping puede estar protegido por derechos de autor. En general, los datos como números y hechos no suelen estar protegidos por derechos de autor, pero la forma en que se presentan o los textos creados por personas sí lo pueden estar. Extraer y reutilizar contenido protegido por derechos de autor sin permiso puede generar problemas legales.

### 4. **Computer Fraud and Abuse Act (CFAA) - Estados Unidos**
En los Estados Unidos, el **Computer Fraud and Abuse Act (CFAA)** regula el acceso no autorizado a computadoras y redes. Algunas demandas de scraping en EE. UU. se han basado en esta ley, alegando que el scraping sin permiso constituye un "acceso no autorizado". Sin embargo, el alcance de la CFAA en relación con el scraping sigue siendo objeto de debate judicial.

### 5. **Consideraciones Éticas**
Además de los aspectos legales, el web scraping plantea consideraciones éticas. Los sitios web a menudo dependen del tráfico para obtener ingresos por publicidad o servicios. El scraping excesivo puede sobrecargar los servidores y causar interrupciones del servicio. Es fundamental hacer un uso responsable de esta técnica, respetando la carga del servidor, la frecuencia de las solicitudes y las políticas de acceso de cada sitio.

### Conclusión

La legalidad del web scraping depende en gran medida de varios factores, incluidos los términos de servicio del sitio, el tipo de datos recopilados y las leyes locales. Para evitar problemas legales, es recomendable:

1. **Revisar los términos de servicio** del sitio web.
2. **Respetar las leyes de protección de datos**, como el RGPD en la UE.
3. **Evitar el scraping de contenido protegido por derechos de autor**.
4. **Limitar la carga en los servidores** y no realizar scraping excesivo.

### Algunos casos judiciales sobre _web scraping_

- **hiQ Labs vs. LinkedIn** (https://en.wikipedia.org/wiki/HiQ_Labs_v._LinkedIn), 2019. La empresa de análisis de datos hiQ Labs extraía información de perfiles públicos de LinkedIn mediante scraping automatizado. Tras recibir un cese-y-desista, hiQ demandó y obtuvo una medida cautelar obligando a LinkedIn a no bloquear su acceso. La Corte de Apelaciones del Noveno Circuito falló a favor de hiQ en 2019, estableciendo que acceder a datos que un sitio pone a disposición del público no constituye acceso "no autorizado" bajo la CFAA. El Tribunal Supremo devolvió el caso en 2021 para su revisión a la luz de Van Buren v. United States, y el Noveno Circuito reafirmó la misma conclusión en 2022.
  
- **Facebook vs. Power Ventures** (https://en.wikipedia.org/wiki/Facebook,_Inc._v._Power_Ventures,_Inc.), 2016. Facebook demandó a Power Ventures, que agregaba cuentas de redes sociales de sus usuarios (incluida Facebook) en un único servicio, por violación de la CFAA. La Corte de Apelaciones del Noveno Circuito falló a favor de Facebook en 2016, estableciendo que acceder a un sistema tras la revocación expresa de la autorización (mediante un cese-y-desista) constituye acceso "no autorizado" bajo la CFAA, con independencia del consentimiento de los usuarios cuyos datos se agregaban.

- **Ryanair vs. Atrápalo** Ryanair vs. Atrápalo (https://www.legaltoday.com/practica-juridica/derecho-civil/nuevas-tecnologias-civil/ryanair-pierde-su-batalla-legal-frente-a-las-agencias-de-viajes-online-2013-01-14/, https://www.lexology.com/library/detail.aspx?g=696747f1-c09c-4f9b-a849-7c4ad6fb13ce), 2012. Ryanair demandó a la agencia de viajes online española Atrápalo por extraer mediante web scraping información de vuelos y precios de su web. El Tribunal Supremo (STS 572/2012, 9 de octubre de 2012) falló a favor de Atrápalo, rechazando que hubiera incumplimiento de los términos de uso, infracción de derechos sobre bases de datos, o competencia desleal. El tribunal consideró el scraping de datos no protegidos una técnica legal.

- **Ryanair Ltd v PR Aviation BV** (https://parissmith.co.uk/tag/ryanair-ltd-v-pr-aviation-bv/), 2015. Este caso fue llevado ante el Tribunal de Justicia de la Unión Europea (TJUE) y es uno de los fallos más importantes en el ámbito del web scraping en Europa. La aerolínea irlandesa Ryanair demandó a la empresa holandesa PR Aviation BV, que utilizaba técnicas de scraping para extraer datos de vuelos y tarifas de Ryanair para ofrecer comparaciones de precios y reservas a través de su propio sitio web sin permiso. El TJUE determinó que los datos de Ryanair no estaban protegidos por la Directiva de Bases de Datos, ya que no cumplían con los requisitos de originalidad o inversión sustancial para ser considerados una base de datos protegida por esa normativa. Sin embargo, el Tribunal falló a favor de Ryanair en cuanto a la validez de sus términos de uso. Aunque los datos no estaban protegidos bajo la Directiva, PR Aviation aún estaba sujeta a los términos y condiciones del sitio web, que prohibían el scraping.

## Web scraping básico: BeautifulSoup

Para realizar *web scraping* la situación más sencilla es disponer de una página estática que se puede acceder a partir de una URL y de la que hay que extraer la información. Una página web HTML es una estructura jerárquica de elementos y necesitamos una manera potente de recorrerla o de buscar elementos concretos. En Python la opción más utilizada es la biblioteca **Beautifulsoup**.

```python
!pip install beautifulsoup4
```

```python
from bs4 import BeautifulSoup

html = """
<!DOCTYPE html>
<html lang="es">
  <head>
    <title>Página de prueba</title>
  </head>
  <body>
    <div>
      <a class="link main" href="https://www.ucm.es">Página UCM</a>
      <a id="deporte_lnk" href="https://www.as.com">As</a>
      <span>
        <a class="link" href="https://www.google.com">Google</a>
      </span>
    </div>
    <a href="http://www.bbva.es">BBVA</a>
  </body>
</html>"""

soup = BeautifulSoup(html, 'html.parser')
```

```python
# Acceso directo a un elemento desde la raíz
enlace = soup.body.div.a
enlace
```

```python
# Acceso a datos de un elemento
print(enlace.name)
print(enlace.attrs)
print('href' in enlace.attrs)
print(enlace['href'])
print(enlace['class'])  # 'class' es un atributo multivaluado
print(enlace.string)
```

```python
# Buscar todos los elementos de un tipo
soup.find_all("a")
```

```python
# Búsqueda de todos los elementos de un tipo con una clase
soup.find_all("a", "link")
```

```python
# Búsqueda de los elementos que cumplen una condición personalizada
def enlace_peligroso(tag):
    return (tag.name == 'a' and 
           'href' in tag.attrs and
            tag['href'].startswith('http://'))

enlaces_peligrosos = soup.find_all(enlace_peligroso)
enlaces_peligrosos
```

```python
# Acceso por ID
soup.find_all(id='deporte_lnk')
```

```python
# Elementos con un determinado href
soup.find_all(href='https://www.ucm.es')
```

```python
# Buscar el primer elemento que cumpla la condición
primer_enlace = soup.find('a')
print(primer_enlace)
print()

primer_span = soup.find('span')
print(primer_span)
print()

primera_imagen = soup.find('img')
print(primera_imagen) # No existe, devuelve None
```

```python
# Acceder al nodo padre
primer_enlace.find_parent()
```

## Ejemplo completo: listado de versiones de Python

Vamos a ver un ejemplo sencillo: acceder a todas las versiones estables de Python. Esa información se encuentra en la página web estática https://www.python.org/downloads/source/. Únicamente estamos interesados en la version (no su fecha) y el enlace a su página. Nos gustaría generar una lista de parejas `(versión, URL)`.

```python
from bs4 import BeautifulSoup
import requests

BASE_URL = 'https://www.python.org'
URL = 'https://www.python.org/downloads/source/'

# Accedemos al código HTML de la página y lo analizamos con bs4
r = requests.get(URL)
soup = BeautifulSoup(r.text, 'html.parser')
```

```python
# Buscamos los divs de tipo columna donde están las versiones de Python
divs = soup.find_all("div", "column")
# divs = soup.find_all("div", class_="column")  # equivalente

# Tiene que haber dos: Stable Releases y Pre-releases
len(divs)
```

```python
# Seleccionamos únicamente con el div que tenga un elemento h2 anidado con texto "Stable Releases"
divs_stable = [div for div in divs if div.find("h2").string == "Stable Releases"]
assert (len(divs_stable) == 1)
div_stable = divs_stable[0]
```

```python
# Filtramos los enlaces cuyo texto empieza por "Python" para descartar enlaces a ficheros comprimidos
# (que empiezan por "Download")
# Para ello crearemos una función que devuelve True solo en este tipo de elementos

def link_starting_python(tag):
    return tag.name == 'a' and tag.string.startswith('Python')

enlaces = div_stable.find_all(link_starting_python)
enlaces
```

```python
# Creo una lista (Versión, fecha, enlace)

def get_version(text):
    [version, date] = text.split( '-' )
    return version.strip() # Elimino espacios innecesarios

def get_year(text):
    [version, date] = text.split( '-' )
    [day_month, year] = date.split(',')
    return int(year.strip()) # Traduzco a entero

lista_python = [(get_version(link.text), get_year(link.text), BASE_URL+link['href']) for link in enlaces]
lista_python[:10]
```

```python
# Ahora podemos consultar fácilmente las versiones de un año concreto, p.ej. 2020
# Las ordenamos de mayor a menor versión

def tupla_version(name):
    """ Extrae la tupla numérica de la versión del nombre
        Ej: 'Python 3.12.6' -> (3, 12, 6)
    """
    version = name[6:]
    version = version.strip()
    version_t = tuple(int(n) for n in version.split('.'))
    return version_t
    
sorted([v[0] for v in lista_python if v[1] == 2024], reverse=True, key=tupla_version)
# Usar las tuplas es imprescindible para ordenar bien, dado que el orden de las cadenas de texto no es adecuado
# Ej: 'Python 3.8.19' > 'Python 3.12.6' como cadenas de texto, pero 
#     (3, 8, 19) < (3, 12, 6) como tuplas   
```
