# Uso básico de los DataFrame de Pandas

Los `DataFrames` son tablas bidimensionales de valores formadas por filas y columnas. Las columnas tienen un nombre y las filas tienen un índice.

## Creación 
Se puede cargar un DataFrame de manera sencilla desde ficheros CSV o Excel.


```python
import pandas as pd

# CSV de datos.gob.es 
# Nº de Médicos por Comunidades, Ciudades autónomas y Provincias de colegiación, edad y sexo
df = pd.read_csv('https://www.ine.es/jaxi/files/tpx/csv_bdsc/59167.csv', sep=';')

# Si la web del INE devuelve error 500, se puede descargar manualmente y abrir desde el disco duro
# df = pd.read_csv('59167.csv', sep=';')

print(df.dtypes)
display(df)
```

También se pueden crear DataFrame a partir de una lista de filas, asignando nombres a las columnas. El caso más sencillo sería el de una tabla de una única columna llamada `'Resultado'`

```python
datos = [5, 3, 5, 1, 4, 6]
df = pd.DataFrame(datos, columns=['Resultado'])
df
```

Como se ve, aunque no hemos indicado nada, automáticamente se ha incluido un índice numérico incremental en las filas.

Si las filas están formadas por varias columnas, los datos serán una lista de listas:

```python
datos = [['España', 'Madrid', 657],
         ['Canadá', 'Ottawa', 70],
         ['Venezuela', 'Caracas', 1037],
         ['Japón', 'Tokio', 6],
         ['Nueva Zelanda','Wellington', 1],
         ['Botsuana', 'Gaborone', 1014]
        ]
df_capitales = pd.DataFrame(datos,
                            columns=['País', 'Capital', 'Altura capital'])
df_capitales
```

Finalmente, podemos indicar el índice de cada fila, que no siempre debe ser un número (en este caso son cadenas de texto)

```python
datos = [['Madrid', 657],
         ['Ottawa', 70],
         ['Caracas', 1037],
         ['Tokio', 6],
         ['Wellington', 1],
         ['Gaborone', 1014]
        ]
df_capitales2 = pd.DataFrame(datos, 
                            columns=['Capital', 'Altura capital'],
                            index=['España', 'Canadá', 'Venezuela', 'Japón','Nueva Zelanda', 'Botsuana'] )
df_capitales2
```
